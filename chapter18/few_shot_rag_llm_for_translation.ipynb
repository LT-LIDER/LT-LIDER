```python
from transformers import pipeline

model_id = "utter-project/EuroLLM-1.7B-Instruct"
pipe = pipeline("text-generation", model=model_id)
```

```python
sentence = "Nothing in this Declaration may be interpreted as implying for any State, group or person any right to engage in any activity or to perform any act aimed at the destruction of any of the rights and freedoms set forth herein."
reference_translation = "Geen bepaling in deze Verklaring zal zodanig mogen worden uitgelegd, dat welke Staat, groep of persoon dan ook, daaraan enig recht kan ontlenen om iets te ondernemen of handelingen van welke aard ook te verrichten, die vernietiging van een van de rechten en vrijheden, in deze Verklaring genoemd, ten doel hebben."
src = 'English'
tgt = 'Dutch'
MAX_NEW_TOKENS = 2 * len(sentence.split())
```

# Zero-shot translation

```python
prompt_content = f"Translate the following text from {src} to {tgt}.\n{src}: {sentence}\n{tgt}:"
messages = [{"role": "user", "content": prompt_content}]
outputs = pipe(messages, return_full_text=False, max_new_tokens=MAX_NEW_TOKENS)
print(outputs[0]['generated_text'])
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
embeddings = embedding_model.encode([source for source, _ in dataset])

rag_sentence = "The night crew had to inspect the points before the first train."
rag_reference = "De nachtploeg moest de wissels inspecteren voordat de eerste trein kwam."
fixed_indices = [1, 2, 6]  # Three general examples, chosen in advance.
scores = embedding_model.similarity(embedding_model.encode(rag_sentence), embeddings)[0]
ranked_indices = scores.argsort(descending=True).tolist()
retrieved_indices = ranked_indices[: len(fixed_indices)]

print("English:", rag_sentence)
print("\nRetrieved examples")
for index in retrieved_indices:
    source, target = dataset[index]
    print(f"{scores[index]:.2f}  {source}\n      {target}")

print("\nExamples used in the fixed prompt")
for index in fixed_indices:
    print(f"{scores[index]:.2f}  {dataset[index][0]}")
```

```python
def translate_with_examples(indices):
    examples = "\n\n".join(
        f"{src}: {dataset[index][0]}\n{tgt}: {dataset[index][1]}" for index in indices
    )
    prompt = (
        f"Translate the final {src} sentence into {tgt}. Return only the translation.\n\n"
        f"{examples}\n\n{src}: {rag_sentence}\n{tgt}:"
    )
    messages = [{"role": "user", "content": prompt}]
    output = pipe(messages, return_full_text=False, do_sample=False, max_new_tokens=64)
    return output[0]["generated_text"].strip()

print("Reference:      ", rag_reference)
print("Fixed few-shot: ", translate_with_examples(fixed_indices))
print("With retrieval: ", translate_with_examples(retrieved_indices))
```

In the saved run, the fixed prompt produced *punten*, and retrieval supplied the railway term *wissels*. Both translations added *vertrok* (“departed”) where the English only says “before the first train.” Check whether that extra detail fits the context.
